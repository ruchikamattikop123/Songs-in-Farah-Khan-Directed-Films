# Dataset checks for Farah Khan film songs

This notebook checks `fk_movie_songs.csv` using pandas. Because each song is a row in a CSV, the file-path check for separate item files does not apply.

```python
from pathlib import Path
import pandas as pd

data_path = Path("fk_movie_songs.csv")
songs = pd.read_csv(data_path)
songs.head()
```

## Number of items

```python
print("Number of items:", len(songs))
```

## Unique IDs

```python
print("All IDs unique:", songs["id"].is_unique)
print("Duplicate IDs:", songs.loc[songs["id"].duplicated(keep=False), "id"].tolist())
```

## Missing values by column

Pandas reads empty CSV cells as missing values.

```python
songs.isna().sum()
```

## Basic summary

Count the listed records per film.

```python
songs.groupby("film_title").size().sort_values(ascending=False)
```

## Possible checks for Future Research

```python
# Count song records with Javed Akhtar listed as a lyricist
lyricists = songs["lyricist"].fillna("").str.split(";").explode().str.strip()

javed_count = (lyricists == "Javed Akhtar").sum()
print("Song records with Javed Akhtar as lyricist:", javed_count)


# Count how many different films list Sunidhi Chauhan as a singer
singer_credits = songs.assign(
    singer=songs["singer"].fillna("").str.split(";")
).explode("singer")

singer_credits["singer"] = singer_credits["singer"].str.strip()

sunidhi_films = sorted(
    singer_credits.loc[
        singer_credits["singer"] == "Sunidhi Chauhan",
        "film_title"
    ].dropna().unique()
)

print("Number of films Sunidhi Chauhan sang in:", len(sunidhi_films))
print("Films:", ", ".join(sunidhi_films))
```
